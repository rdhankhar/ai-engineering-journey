```python
# String Handling

df["gender"].str.upper()
df["gender"].str.lower()
df["gender"].str.capitalize()

df["name"].str.strip()

df["name"].str.split(" ")

df["country"].str.contains("india", case = False)
```

```python
# Data Transformation

d2 = df.copy()

#Apply

d2["tax"] = d2["income"].apply([lambda x: "20%" if x >= 60000 else "10%"])

#Assign

d2 = d2.assign(new_income = d2["income"] * 1.1)

# Map

gender_map = {"Male" : "M", "Female" : "F"}

d2["gender"] = d2["gender"].map(gender_map)

# Replace

d2["country"] = d2["country"].replace("USA" , "United States")

# Rename

# d2.columns = ["ID","NAME","AGE","COUNTRY","GENDER","INCOME","TAX","NEW_INCOME"]
d2.rename(columns={"name" : "NAME"})
d2.rename(index={1:"First"})

# sort_values
d2.sort_values("income")
d2["age"] = d2["age"].fillna(50)
sorted_values = d2.sort_values(["income","age"] , ascending = False)

sorted_values
# sort_index

sorted_values.sort_index()

# reset_index

sorted_values.reset_index()

# Ranking

sorted_values["ranking"] = sorted_values["income"].rank(ascending = True, method ="min")

sorted_values[["id","name","age","country","gender","income","new_income","tax"]]

sorted_values
```
