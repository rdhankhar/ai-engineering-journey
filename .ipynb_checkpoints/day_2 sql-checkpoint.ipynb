```python
import pandas as pd
from pandasql import sqldf
df = pd.read_csv(r"C:\Users\ASUS\DA Journey\train.csv")

pysql = lambda q: sqldf(q, globals())

print("dataset Loaded! shape :", df.shape)
```

```python
result = pysql("SELECT * FROM df LIMIT 10")
result
```

```python
result = pysql("""
select Sex ,COUNT(*) as train_passengers
FROM df 
GROUP BY Sex """)
result
```

```python
result = pysql("""
SELECT Name , Sex , Age , Pclass , Fare 
FROM df 
where Fare > 100 
ORDER BY Fare DESC 
LIMIT 10
""")
result
```

```python
result = pysql("""
SELECT Pclass , COUNT(*) as total_passengers,
ROUND(AVG(Fare),2) as avg_fare
FROM df 
GROUP BY Pclass
ORDER BY Pclass """)
result

```
