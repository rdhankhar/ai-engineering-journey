```python
# plot customization and intro 


import matplotlib.pyplot as plt

import numpy as np
x = np.array([2020 , 2023 , 2025 , 2022, 2021])

y1 = np.array([ 20, 30 ,35 , 40 , 90])
y2 = np.array([ 23, 32 ,38 , 43 , 97])
y3 = np.array([ 15, 50 ,55 , 60,  85])

line_style = dict(marker = ".",
         markersize = 25,
         markerfacecolor = "#42f5ad",
         markeredgecolor = "#42f5ad",
         linestyle = "solid",
         linewidth = 4 )
# plt.plot(x,y, marker = ".",
#          markersize = 25,
#          markerfacecolor = "#42f5ad",
#          markeredgecolor = "#42f5ad",
#          linestyle = "dashed")
#          linestyle = "dotted")
#          linestyle = "dashdot")
#          linestyle = "none")
#          linestyle = "solid",
#          linewidth = 4 )

plt.plot(x, y1,color = "#4542f5" , ** line_style)
plt.plot(x, y2,color = "#4542f5" , ** line_style)
plt.plot(x, y3,color = "#4542f5" , ** line_style)

# plt.plot(y)

plt.show()
```
